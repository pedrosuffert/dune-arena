```python
# Compute the Macro average
classes = ['normal', 'ransomware', 'ddos', 'xss', 'scanning', 'injection', 'password']
python.loc['f1-score', 'macro avg'] = python.loc['f1-score', classes].mean()
```

```python
python.loc[('f1-score', 'support'), classes]
```

```python
# Compute the Weighted average
python.loc['f1-score', 'weighted avg'] = (python.loc[('f1-score'), classes] * python.loc[('support'), classes]).sum() / python.loc[('support'), classes].sum()
```

```python
# Compute the Micro average
## Initialize totals
TP_total = FP_total = FN_total = 0.0

## Recover TP, FP, FN per class
for key,metrics in python_results.items():
    if key in exclude_cols:
        print(f'excluding: {key}')
        continue
    print(f'computing: {key}')
    precision = metrics['precision']
    recall = metrics['recall']
    support = metrics['support']
    
    TP = recall * support
    FP = TP * (1/precision - 1)
    FN = support - TP
    
    TP_total += TP
    FP_total += FP
    FN_total += FN

## Compute micro F1
micro_f1 = 2 * TP_total / (2 * TP_total + FP_total + FN_total)
print("Micro F1:", micro_f1)

python.loc['f1-score', 'micro avg'] = micro_f1
```
